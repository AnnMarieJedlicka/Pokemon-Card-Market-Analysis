```python
import streamlit as st
import pandas as pd
import plotly.express as px

st.title("Pokemon Cards Market Dashboard")

st.metric(
    "Average Price",
    round(df["price_usd"].mean(),2)
)

fig = px.bar(
    df.groupby("rarity_class")["price_usd"]
    .mean()
    .reset_index(),
    x="rarity_class",
    y="price_usd"
)

st.plotly_chart(fig)
```

Chart 1: Top 10 Sets by Average Price

```python
import matplotlib.pyplot as plt

top_sets = (
    df.groupby("set_name")["price_usd"]
    .mean()
    .sort_values(ascending=False)
    .head(10)
)

top_sets.plot(kind="bar")

plt.title('Top 10 Sets by Average Price')
plt.ylabel('Average Price (USD)')
plt.xlabel('Set')
plt.xticks(rotation=45)
plt.tight_layout()

plt.savefig("top10_sets.png")
```

Chart 2: Average Price by Rarity

```python
rarity_avg = (
    df.groupby('rarity_class')['price_usd']
    .mean()
    .sort_values(ascending=False)
)

rarity_avg.plot(kind='bar')

plt.title('Average Price by Rarity')
plt.ylabel('Average Price (USD)')
plt.xlabel('Rarity')
plt.tight_layout()

plt.show()
```

Chart 5: Top Pokemon by Total Market Value

```python
top_total = (
    df.groupby('pokemon_name')['price_usd']
      .sum()
      .sort_values(ascending=False)
      .head(10)
)

top_total.plot(
    kind='bar',
    figsize=(10,6),
    color='darkred'
)

plt.title('Top Pokemon by Total Market Value')
plt.xlabel('Pokemon')
plt.ylabel('Total Value (USD)')
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

Chart 6: Top 10 Pokemon by Average Market Value

```python
import matplotlib.pyplot as plt

top_pokemon = (
    df.groupby('pokemon_name')['price_usd']
      .mean()
      .sort_values(ascending=False)
      .head(10)
)

ax = top_pokemon.plot(
    kind='bar',
    figsize=(10,6),
    color='royalblue'
)

plt.title('Top 10 Pokemon by Average Market Value')
plt.xlabel('Pokemon')
plt.ylabel('Average Price (USD)')
plt.xticks(rotation=45)

for container in ax.containers:
    ax.bar_label(container, fmt='$%.0f')

plt.tight_layout()
plt.show()
```
